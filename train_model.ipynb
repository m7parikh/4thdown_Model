Loads the cleaned 4th-down dataset from `pull_data.ipynb` and builds a model that predicts EPA (expected points added) for a given game situation and decision.

```python
import pandas as pd
```

```python
fourth = pd.read_parquet("data/fourth_downs.parquet")
fourth.shape
```

## EPA as the decision-quality label

We use `epa` — already computed by nflverse for every play — as the label for "how good was this decision," instead of building our own expected-points model. It reflects the full outcome of the play, including the risk of turning it over on downs when going for it.

Sanity-check the direction before trusting it as a target: going for it on short yardage should show a *higher* average EPA than punting or kicking, since that's the well-documented finding in 4th-down research.

```python
print("epa nulls:", fourth["epa"].isna().sum())
```

```python
print("mean epa by decision (all situations):")
print(fourth.groupby("decision")["epa"].mean().sort_values(ascending=False))
```

```python
short = fourth[fourth["ydstogo"] <= 1]
print(f"mean epa by decision, 4th-and-1 or less (n={len(short):,}):")
print(short.groupby("decision")["epa"].mean().sort_values(ascending=False))
```
